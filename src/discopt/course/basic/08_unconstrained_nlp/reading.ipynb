## 1. Optimality conditions

For unconstrained $\min_x f(x)$ with $f \in C^2$:

- **First-order:** $\nabla f(x^\star) = 0$ (necessary).
- **Second-order:** $\nabla^2 f(x^\star) \succeq 0$ (necessary), $\succ 0$ sufficient for strict local min.

These are necessary, not sufficient — saddle points satisfy first-order. {cite:p}`Nocedal2006` Ch 2.

## 2. Gradient descent

$$x_{k+1} = x_k - \alpha_k \nabla f(x_k).$$

- Cheap (just need gradients).
- Linear convergence on strongly convex $f$ with condition number $\kappa$: error shrinks by $(1 - 1/\kappa)$ each step. Slow when $\kappa$ is large.
- Step $\alpha_k$: line search (Armijo, Wolfe) or fixed $\alpha = 1/L$ if you know the Lipschitz constant {cite:p}`Nocedal2006`.

## 3. Newton's method

$$x_{k+1} = x_k - [\nabla^2 f(x_k)]^{-1} \nabla f(x_k).$$

- **Quadratic convergence** in a neighbourhood of $x^\star$ when $\nabla^2 f(x^\star)$ is *positive definite* (not merely PSD) and the Hessian is locally Lipschitz {cite:p}`Nocedal2006` (Thm 3.5). Without the PD hypothesis, convergence can drop to linear — e.g., for $f(x) = x^4$ the Hessian vanishes at the optimum and Newton gives $x_{k+1} = \frac{2}{3} x_k$.
- Requires Hessians (and inverting them).
- Far from optimum, can diverge — modified Newton with damping or trust region {cite:p}`Nocedal2006`.

## 4. Quasi-Newton (BFGS)

Build an approximate Hessian $B_k$ from gradient differences:

$$B_{k+1} = B_k - \frac{B_k s_k s_k^\top B_k}{s_k^\top B_k s_k} + \frac{y_k y_k^\top}{y_k^\top s_k},$$

with $s_k = x_{k+1} - x_k$, $y_k = \nabla f(x_{k+1}) - \nabla f(x_k)$. Superlinear convergence; no Hessian needed; the workhorse for unconstrained NLP.

```python
import numpy as np

# Rosenbrock — non-convex but a unique global min at (1,1).
# We compute gradients with JAX autodiff here. (discopt builds its own
# derivatives -- by default from a Rust automatic-differentiation tape --
# so its solvers get exact gradients and Hessians of your model for free.)
import jax
import jax.numpy as jnp

def f(x): return (1 - x[0])**2 + 100 * (x[1] - x[0]**2)**2
grad_f = jax.grad(lambda x: f(x))

x = np.array([-1.2, 1.0])
for k in range(50):
    g = np.asarray(grad_f(jnp.array(x)))
    x = x - 0.001 * g
print("after 50 GD steps: x =", x, "  f =", f(x))
# (Far from the optimum (1,1) — GD is slow on Rosenbrock.)
```

## 5. Line search

We want $\alpha_k$ such that $f(x_k + \alpha_k p_k)$ decreases sufficiently. Wolfe conditions:

- $f(x + \alpha p) \le f(x) + c_1 \alpha \nabla f^\top p$ (Armijo / sufficient decrease).
- $|\nabla f(x + \alpha p)^\top p| \le c_2 |\nabla f^\top p|$ (curvature).

Practical choice $c_1 = 10^{-4}, c_2 = 0.9$. See {cite:p}`Nocedal2006` Ch 3.

## 6. Trust region

Solve a *local quadratic model* in a ball $\|p\| \le \Delta$:

$$\min_p \; m_k(p) = f_k + \nabla f_k^\top p + \tfrac{1}{2} p^\top B_k p \;\;\text{s.t.}\;\; \|p\| \le \Delta.$$

Adapt $\Delta$ based on agreement between $m_k$ and $f$. More robust than line search when Hessian has indefinite directions.

{cite:p}`Nocedal2006` is the standard reference. Also {cite:p}`Wright1997`.

## 7. What `discopt` runs

`discopt` never asks you to pick GD vs Newton. Every continuous NLP solve goes to **POUNCE**, a Rust port of IPOPT {cite:p}`Wachter2006`: a *Newton* method on the KKT conditions using exact Hessians from automatic differentiation, globalized by a **filter line search** (Lesson 9), with Hessian regularization when the Newton direction is not a descent direction — exactly the "pure Newton can fail" fixes from Sections 3 and 5. For a nonconvex objective like Rosenbrock, the default `m.solve()` wraps those local solves in spatial branch-and-bound to *certify* the global minimum, which costs thousands of nodes here. Ask for a single *local* solve with `m.solve(skip_convex_check=True, initial_solution=...)` and you see the Newton method alone: `r.solve_report` holds POUNCE's structured report, including `statistics["iteration_count"]` and a per-iteration trajectory (`iterations`, with objective, step size `alpha_primal` and Hessian `regularization` for each step). Compare its iteration count with your gradient descent from Exercise 1.

When you have no derivatives at all — a black-box simulation wrapped with `dm.custom(...)` — `m.solve(solver="direct")` runs the derivative-free DIRECT global search over the variable box (`max_evals` controls the effort) and `solver="surrogate"` fits a surrogate model; neither returns a certificate (`r.gap_certified` is `False`).

```python
import logging
import numpy as np
import discopt as do

# DIRECT never certifies, and discopt logs a warning saying so; silence it here.
logging.getLogger("discopt").setLevel(logging.ERROR)

m = do.Model("rosenbrock")
xv = m.continuous("x", shape=(2,), lb=-5, ub=5)
m.minimize((1 - xv[0])**2 + 100 * (xv[1] - xv[0]**2)**2)

# One local Newton-type (interior-point) solve from the classic start (-1.2, 1).
r = m.solve(skip_convex_check=True, initial_solution={xv: np.array([-1.2, 1.0])})
rep = r.solve_report
print("local POUNCE:", r.status, " x =", r.value(xv).round(6),
      " iterations:", rep["statistics"]["iteration_count"])
print("  full steps (alpha = 1):", sum(it["alpha_primal"] == 1.0 for it in rep["iterations"]),
      " objective by iteration:", [f"{it['objective']:.1e}" for it in rep["iterations"][::4]])

r = m.solve(solver="direct", max_evals=3000)      # derivative-free, no certificate
print("DIRECT:", r.status, " x =", r.value(xv).round(6), " f =", f"{r.objective:.2e}",
      " certified:", r.gap_certified)
```
