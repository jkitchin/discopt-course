## 1. The two-stage SP

$$\min_x c^\top x + \mathbb{E}_\xi[Q(x, \xi)] \quad \text{s.t.} \quad x \in \mathcal{X}$$

where $Q(x, \xi) = \min_y \{q^\top y : Wy = h - Tx, y \ge 0\}$ is the second-stage value under realization $\xi = (q, h, T, W)$.

If $\xi$ has finitely many scenarios $s = 1, \dots, S$ with probabilities $p_s$, the **extensive form** is one big LP. With many scenarios, decomposition (Benders, Lesson 18) is essential. {cite:p}`Birge2011` is the standard reference.

## 2. Sample average approximation (SAA)

For continuous $\xi$, sample $N$ realizations and solve the *deterministic* SP with these $N$ scenarios. As $N \to \infty$, the optimal value converges (SAA consistency). {cite:p}`Shapiro2009` Ch 5.

In practice: sample 100, 500, 1000 scenarios; solve; compare optimal values to gauge stability.

## 3. Chance constraints

$$\Pr_\xi(\xi^\top x \le b) \ge 1 - \epsilon.$$

For Gaussian $\xi \sim \mathcal{N}(\bar\xi, \Sigma)$ and $\epsilon \le 1/2$, this is equivalent to a deterministic constraint:

$$\bar \xi^\top x + \Phi^{-1}(1 - \epsilon) \,\|\Sigma^{1/2} x\|_2 \le b,$$

an SOCP constraint, since $\Phi^{-1}(1-\epsilon) \ge 0$. (For $\epsilon > 1/2$ the set is nonconvex.) For non-Gaussian, harder; SAA + integer indicators is one way. {cite:p}`Birge2011`.

## 4. CVaR

For a confidence level $\alpha$ (e.g. $0.95$), $\text{VaR}_\alpha(L)$ is the $\alpha$-quantile of the loss, and for a continuous loss distribution

$$\text{CVaR}_\alpha(L) = \mathbb{E}[L \mid L \ge \text{VaR}_\alpha(L)],$$

the mean of the worst $(1-\alpha)$ fraction of outcomes. The Rockafellar–Uryasev formula $\text{CVaR}_\alpha(L) = \min_t \{ t + \mathbb{E}[(L - t)^+]/(1-\alpha) \}$ {cite:p}`RockafellarUryasev2000` holds for any distribution, and turns CVaR over scenarios into an LP. For a random loss $L(x, \xi)$, minimizing CVaR is convex in $x$ when $L$ is convex. Replaces the non-convex VaR; widely used in risk-aware optimization.

```python
# Standard discopt course imports. The lessons target the real
# `discopt.modeling.core.Model` API: `m.continuous(name, shape=..., lb=..., ub=...)`,
# `m.binary(name, shape=...)`, `m.integer(name, shape=..., lb=..., ub=...)`,
# `m.subject_to(...)`, `m.minimize(...) / .maximize(...)`, `m.solve(...)`.
# Result attributes: `r.status`, `r.objective`, `r.gap`, `r.bound`,
# `r.wall_time`, `r.node_count`, and `r.value(var)` for variable arrays.
import numpy as np
import discopt as do
import discopt.modeling as dm
```

```python
import numpy as np, discopt as do

# Newsvendor: order x today, demand D ~ uniform(50, 150)
# Cost: 2*x today, revenue 5*min(x, D), salvage 1*max(0, x - D)
np.random.seed(0)
N = 200
D = np.random.uniform(50, 150, size=N)

m = do.Model("newsvendor")
x = m.continuous("x", lb=0, ub=200)
y = m.continuous("y", shape=(N,), lb=0, ub=200)     # sold (never more than x <= 200)
z = m.continuous("z", shape=(N,), lb=0, ub=200)     # salvage
for s in range(N):
    m.subject_to(y[s] <= x)
    m.subject_to(y[s] <= D[s])
    m.subject_to(z[s] == x - y[s])      # equality! z = unsold = salvage quantity
# Salvage z is recovered revenue (1 per unit), NOT a cost — both 5*y and 1*z carry the same sign.
m.minimize(2 * x - sum(5 * y[s] + 1 * z[s] for s in range(N)) / N)
r = m.solve()
print(f"order quantity: {r.value(x):.2f}, expected profit: {-r.objective:.2f}")
```

## 5. Multi-stage SP

For $T$ stages, decisions are recourse functions $x_t(\xi_{1:t})$. Number of scenarios grows exponentially in $T$. Approximation strategies:

- **Scenario tree** (small finite tree).
- **SDDP** (stochastic dual dynamic programming) — Benders cuts in time {cite:p}`Birge2011`.

Multi-stage SP is the bread and butter of energy and finance applications.

## 6. `discopt.stochastic`

`discopt` ships a two-stage stochastic-programming layer in `discopt.stochastic`:

- `ScenarioSet.from_list([(p, data), ...])` / `ScenarioSet.from_samples({"demand": array})` (SAA, equal weights);
- `build_extensive_form(m, scenarios=..., recourse_builder=f, first_stage_cost=..., risk=...)`. This calls `f(model, data, s)` once per scenario to add that scenario's recourse variables and constraints, and returns its cost expression. It then writes the probability-weighted objective. `risk=Expectation()` (default), `CVaR(alpha)`, or `MeanCVaR(...)`;
- `solve_lshaped(...)`: the L-shaped method {cite:p}`VanSlykeWets1969`, i.e. Benders on the same scenario structure;
- `progressive_hedging(...)` {cite:p}`RockafellarWets1991`, `chance_constraint(...)` (SAA with big-M indicators), and `saa_lower_bound_estimate` / `saa_upper_bound_estimate` / `optimality_gap_estimate` for the statistical bounds of {cite:t}`MakMortonWood1999`.

*Not* in `discopt` yet: multistage SP (`solve_multistage` raises `NotImplementedError`; nested Benders/SDDP is planned) and integer L-shaped. For those, write the deterministic equivalent by hand, as in Exercise 4.

The cell below rebuilds the newsvendor from §4 with this layer and compares the risk-neutral order with a CVaR-averse one.

## References
{cite:p}`Birge2011,Shapiro2009,Wolsey1998`.

```python
from discopt.modeling.core import Constant
from discopt.stochastic import ScenarioSet, build_extensive_form, Expectation, CVaR

scen = ScenarioSet.from_samples({"demand": D})        # the same N = 200 samples as above

def solve_newsvendor(risk):
    m = do.Model("newsvendor_sp")
    x = m.continuous("x", lb=0, ub=200)                # first stage: order quantity

    def recourse(model, data, s):                      # one call per scenario
        d = float(data["demand"])
        sold = model.continuous(f"sold_{s}", lb=0, ub=200)
        model.subject_to(sold <= x)
        model.subject_to(sold <= Constant(d))
        # cost = -(revenue + salvage of the unsold x - sold)
        return Constant(-5.0) * sold + Constant(-1.0) * (x - sold)

    build_extensive_form(m, scenarios=scen, recourse_builder=recourse,
                         first_stage_cost=Constant(2.0) * x, risk=risk)
    return m.solve()

for label, risk in [("expectation", Expectation()), ("CVaR(alpha=0.9)", CVaR(alpha=0.9))]:
    rs = solve_newsvendor(risk)
    print(f"{label:16s} order = {float(rs.x['x']):7.2f}   objective = {rs.objective:9.3f}")
```
